This notebook trains the YOLO model for ball detection. 

Requires a Roboflow API key in a `.env` file at the top level of this repository. The resulting model is stored within a folder in `courtvision-engine` called `models`.

```python
from dotenv import load_dotenv
from roboflow import Roboflow
import os
import shutil

import torch

load_dotenv(dotenv_path = "../../.env")
api_key = os.getenv("API_KEY")

device = (
    "cuda" if torch.cuda.is_available() 
    else "mps" if torch.backends.mps.is_available() 
    else "cpu"
)
```

```python
# Access Dataset
rf = Roboflow(api_key = api_key)
project = rf.workspace("viren-dhanwani").project("tennis-ball-detection")
version = project.version(6)
dataset = version.download("yolov5", location = "../data/ball_detection_data")
```

```python
!yolo task=detect mode=train model=yolov5l6u.pt data={dataset.location}/data.yaml epochs=100 imgsz=640 device={device}
```
